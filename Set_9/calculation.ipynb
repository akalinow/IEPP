### Constants and parameters

```python
import numpy as np
import scipy.constants as const
from termcolor import colored

m_e, unit_m_e, uncertainty = const.physical_constants['electron mass energy equivalent in MeV']
m_p, unit_m_p, _ = const.physical_constants['proton mass energy equivalent in MeV']
q_ele, unit_q_ele, _ = const.physical_constants['elementary charge']
c, unit_c, _ = const.physical_constants['speed of light in vacuum']
epsilon0, unit_epsilon0, _ = const.physical_constants['vacuum electric permittivity']
rho = 2803 # m
dEdx = 5 # MeV/m
beta = 1.0
```

### Formulas for energy loss per turn:

\begin{align*}
  & E_{L} = P_{L} \cdot \frac{2\pi \rho}{\beta c} = \frac{e^{2}}{3 \epsilon_0} \frac{\rho}{m^{2} c^{4} \beta}  \left( \frac{dE}{dx} \right)^{2} \\
  & E_{T} = P_{T} \cdot \frac{2\pi \rho}{\beta c} = \frac{e^{2}}{3\epsilon_0} \frac{\gamma^{4} \beta^{3} c}{\rho}
\end{align*}

```python
def lossPerTurn(m, gamma):
    e_L = (q_ele**2/(3*epsilon0)) * rho/(m**2*c**4 * beta) * dEdx**2
    e_T = (q_ele**2/(3*epsilon0)) * (gamma**4 * beta**3) * (1/rho)
    #units: J-> MeV
    e_L = e_L/q_ele/1E6
    e_T = e_T/q_ele/1E6
    return e_L, e_T


e_L_ele, e_T_ele = lossPerTurn(m_e/c**2, 100E3/m_e)
e_L_p, e_T_p = lossPerTurn(m_p/c**2, 6500E3/m_p)



print(colored("Electron loss:", "blue"),f" E_L = {e_L_ele:.3e} MeV, E_T = {e_T_ele:.3e} MeV")
print(colored("Proton loss:", "red"),f" E_L = {e_L_p:.3e} MeV, E_T = {e_T_p:.3e} MeV")
```
